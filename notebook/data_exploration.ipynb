```python
from pathlib import Path

project_dir = Path.cwd().parent

file_path = (
    project_dir
    / "data"
    / "processed"
    / "extracted_documents.jsonl"
)

print("File exists:", file_path.exists())
```

```python
import json
import pandas as pd

records = []

with file_path.open("r", encoding="utf-8") as file:
    for line in file:
        records.append(json.loads(line))

df = pd.DataFrame(records)

print("Dataset shape:", df.shape)
df.head()
```

```python
import json
import pandas as pd
from pathlib import Path

project_dir = Path.cwd().parent

cleaned_file = (
    project_dir
    / "data"
    / "processed"
    / "cleaned_documents.jsonl"
)

records = []

with cleaned_file.open(
    "r",
    encoding="utf-8"
) as file:

    for line in file:
        records.append(json.loads(line))

clean_df = pd.DataFrame(records)

print("Cleaned dataset shape:", clean_df.shape)

clean_df.head()
```

```python
clean_df[

    ["title", "word_count", "character_count"]

].sort_values(

    by="word_count",

    ascending=True

)
```

```python
clean_df[

    ["source_url", "title", "text"]

].isnull().sum()
```

```python
import json
import pandas as pd
from pathlib import Path

project_dir = Path.cwd().parent

chunks_file = (
    project_dir
    / "data"
    / "processed"
    / "chunks.jsonl"
)

chunks = []

with chunks_file.open(
    "r",
    encoding="utf-8"
) as file:
    for line in file:
        chunks.append(json.loads(line))

chunks_df = pd.DataFrame(chunks)

print("Chunk dataset shape:", chunks_df.shape)

chunks_df.head()
```
